```python
# 1. Import Libraries
import pandas as pd
import numpy as np

# 2. Load Dataset
df = pd.read_csv('/content/titanic.csv')
```

```python
# 3. Dataset Structure
print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
```

```python
# 4. Dataset Information
print("\nDataset Information:")
df.info()
```

```python
# 5. First 5 Records
print("\nFirst 5 Records:")
display(df.head())
```

```python
# 6. Statistical Overview
print("\nStatistical Summary:")
display(df.describe(include='all'))
```

```python
# 1. Check Missing Values
print("Missing Values:")
display(df.isnull().sum())

# 2. Check Duplicate Records
print("\nDuplicate Records:", df.duplicated().sum())
```

```python
# 1. Handle Missing Values

# Fill Age with Median
df['Age'] = df['Age'].fillna(df['Age'].median())

# Fill Fare with Median
df['Fare'] = df['Fare'].fillna(df['Fare'].median())

# Fill Cabin with Unknown
df['Cabin'] = df['Cabin'].fillna('Unknown')


# 2. Handle Inconsistent Data

df['Sex'] = df['Sex'].str.strip().str.lower()

df['Embarked'] = df['Embarked'].str.strip().str.upper()


# 3. Check and Correct Data Types

df['Age'] = pd.to_numeric(df['Age'])
df['Fare'] = pd.to_numeric(df['Fare'])


# 4. Verify Data Cleaning

print("Missing Values After Cleaning:")
display(df.isnull().sum())

print("\nDuplicate Records:", df.duplicated().sum())

print("\nUpdated Data Types:")
print(df.dtypes)

print("\nCleaned Dataset:")
display(df.head())
```

```python
# 3. Identify Data Types
print("\nColumn Data Types:")
display(df.dtypes)
```

```python
# 4. Separate Numerical and Categorical Columns
numerical_columns = df.select_dtypes(include=np.number).columns.tolist()
categorical_columns = df.select_dtypes(include='object').columns.tolist()

print("\nNumerical Columns:", numerical_columns)
print("\nCategorical Columns:", categorical_columns)

```

```python
# 5. Python Data Structures

# List
columns_list = list(df.columns)

# Tuple
dataset_shape = tuple(df.shape)

# Dictionary
dataset_details = {
    "Rows": df.shape[0],
    "Columns": df.shape[1],
    "Dataset": "Titanic"
}

# Set
unique_passenger_classes = set(df['Pclass'].dropna())

print("\nPython Data Structures:")
print("List:", columns_list)
print("Tuple:", dataset_shape)
print("Dictionary:", dataset_details)
print("Set:", unique_passenger_classes)

```

```python
# 6. Conditional Statements and Loops

for column in df.columns:
    if df[column].isnull().sum() > 0:
        print(column, "contains missing values")
    else:
        print(column, "has no missing values")
```

```python
# 7. User Defined Function

def calculate_survival_rate(data):
    return data['Survived'].mean() * 100

survival_rate = calculate_survival_rate(df)

print("\nOverall Survival Rate:", round(survival_rate, 2), "%")
```

```python
# Select Important Numerical Columns
numeric_data = df[['Age', 'Fare', 'SibSp', 'Parch', 'Pclass', 'Survived']]

# 1. Mean
print("MEAN:")
display(numeric_data.mean())

# 2. Median
print("\nMEDIAN:")
display(numeric_data.median())

# 3. Mode
print("\nMODE:")
display(numeric_data.mode())

# 4. Standard Deviation
print("\nSTANDARD DEVIATION:")
display(numeric_data.std())

# 5. Variance
print("\nVARIANCE:")
display(numeric_data.var())

# 6. Correlation
print("\nCORRELATION:")
display(numeric_data.corr().round(2))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(9, 6))

sns.heatmap(
    numeric_data.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap - Titanic Dataset')
plt.show()
```

```python
plt.figure(figsize=(15, 10))

# 1. Survival Distribution
plt.subplot(2, 2, 1)
sns.countplot(data=df, x='Survived')
plt.title('Passenger Survival Distribution')
plt.xlabel('Survived (0 = No, 1 = Yes)')

# 2. Age Distribution
plt.subplot(2, 2, 2)
sns.histplot(data=df, x='Age', bins=20, kde=True)
plt.title('Age Distribution')

# 3. Survival by Gender
plt.subplot(2, 2, 3)
sns.barplot(data=df, x='Sex', y='Survived')
plt.title('Survival Rate by Gender')

# 4. Survival by Passenger Class
plt.subplot(2, 2, 4)
sns.barplot(data=df, x='Pclass', y='Survived')
plt.title('Survival Rate by Passenger Class')

plt.tight_layout()
plt.show()
```

```python
# Survival Rate by Gender
print("Survival Rate by Gender:")
display(df.groupby('Sex')['Survived'].mean() * 100)

# Survival Rate by Passenger Class
print("\nSurvival Rate by Passenger Class:")
display(df.groupby('Pclass')['Survived'].mean() * 100)

# Survival Rate by Age Group
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[0, 12, 18, 35, 60, 100],
    labels=['Child', 'Teenager', 'Young Adult', 'Adult', 'Senior']
)

print("\nSurvival Rate by Age Group:")
display(df.groupby('Age_Group', observed=False)['Survived'].mean() * 100)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.boxplot(y=df['Age'])
plt.title('Age Outlier Detection')

plt.subplot(1, 2, 2)
sns.boxplot(y=df['Fare'])
plt.title('Fare Outlier Detection')

plt.tight_layout()
plt.show()
```

```python
for column in ['Age', 'Fare']:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    print(f"\n{column} Outliers:")
    print("Number of Outliers:", len(outliers))
```

```python
# Gender-wise Passenger Count
print("Passenger Count by Gender:")
display(df['Sex'].value_counts())

# Gender-wise Survival Count
print("\nGender and Survival Cross-Tabulation:")
display(pd.crosstab(df['Sex'], df['Survived']))

# Gender-wise Survival Percentage
print("\nGender-wise Survival Percentage:")
display(pd.crosstab(
    df['Sex'],
    df['Survived'],
    normalize='index'
) * 100)
```
